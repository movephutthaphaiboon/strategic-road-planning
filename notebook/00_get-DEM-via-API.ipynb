# Download DEM via API Key

```python
import os
import requests

API_KEY = "0e09a52918c555ed6e7e3874db10d82d"   # get from https://portal.opentopography.org/requestService?service=api
OUT_DIR = "dem_tiles"
os.makedirs(OUT_DIR, exist_ok=True)
```

```python
# (west, south, east, north) or (min_x, min_y, max_x, max_y)
tiles = {
    # "N00E006": (6, 0, 9, 3),
    # "N00E009": (9, 0, 12, 3),
    # "N00E012": (12, 0, 15, 3),
    # "N00E015": (15, 0, 18, 3),
    # "N03E006": (6, 3, 9, 6),
    # "N03E009": (9, 3, 12, 6),
    # "N03E012": (12, 3, 15, 6),
    # "N03E015": (15, 3, 18, 6),
    # "N06E006": (6, 6, 9, 9),
    # "N06E009": (9, 6, 12, 9),
    # "N06E012": (12, 6, 15, 9),
    "N06E015": (15, 6, 18, 9),
    # "N09E006": (6, 9, 9, 12),
    # "N09E009": (9, 9, 12, 12),
    # "N09E012": (12, 9, 15, 12),
    "N09E015": (15, 9, 18, 12),
    # "N12E006": (6, 12, 9, 15),
    # "N12E009": (9, 12, 12, 15),
    # "N12E012": (12, 12, 15, 15),
    # "N12E015": (15, 12, 18, 15),
}
```

```python
BASE_URL = "https://portal.opentopography.org/API/globaldem"

for name, (west, south, east, north) in tiles.items():
    params = {
        "demtype": "COP30",      # Copernicus 30 m DEM
        "west": west,
        "south": south,
        "east": east,
        "north": north,
        "outputFormat": "GTiff", # GeoTIFF
        "API_Key": API_KEY,
    }

    print(f"Requesting {name} ...")
    r = requests.get(BASE_URL, params=params, timeout=600)

    if r.status_code == 200:
        out_path = os.path.join(OUT_DIR, f"COP30_{name}.tif")
        with open(out_path, "wb") as f:
            f.write(r.content)
        print(f"Saved {out_path}")
    else:
        print(f"Failed {name}: {r.status_code} {r.text}")
```
