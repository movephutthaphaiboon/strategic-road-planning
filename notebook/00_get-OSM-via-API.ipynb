# Cameroon Roads

## Download

```python
import osmnx as ox

# Replace "Macaroon" with the actual place name if needed
# Example: "Paris, France" or "New York, USA"
place_name = "Cameroon"

# Download road network (all drivable roads)
G = ox.graph_from_place(place_name, network_type='drive')

# Convert to GeoDataFrame for further analysis or export
edges = ox.graph_to_gdfs(G, nodes=False)

# Display the first few rows
edges.head()
```

```python
edges.to_file("macaroon_roads.geojson", driver="GeoJSON")
```
